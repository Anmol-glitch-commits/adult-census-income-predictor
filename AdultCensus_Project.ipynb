### **Outlier for all the numerical Individual features**

**So , Here when i have seen the outier value the values are some extreme observations but they are legitimate outliers...In this the Age is not something like -(something) or hours_per_week = 5000 rather it is 99..so the observation are meaningul..so i will simply keep these outliers.rather than removal**

```python
num_cols = df.select_dtypes(include="number").columns

for col in num_cols:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75) # Corrected Q3 to be 0.75 for IQR calculation

  IQR = Q3 - Q1

  lower_bound = Q1 - 1.5*IQR
  upper_bound = Q3 + 1.5*IQR

  #see the outliers val..
  outliers = df[
    (df[col] < lower_bound) | (df[col] > upper_bound)
  ]

  #now , print them:
  print(col)
  print("Lower Bound: " , lower_bound)
  print("Upper Bound: " , upper_bound)
  print("Outliers: " , len(outliers))
  print()
```

Age shows a noticeable relationship with income in this dataset. Individuals earning more than 50K have a higher average age (44.25 years) compared with those earning 50K or less (36.78 years), and the median age also differs considerably (44 vs. 34 years). The box plot further confirms that the >50K income group is shifted toward higher ages. Although the IQR method identifies observations around 75–90 years as potential outliers, these represent realistic human ages rather than obvious data-entry errors. Therefore, these observations should not be removed solely because they are classified as statistical outliers.

### Model Training

```python
X = df.drop(columns = ['income'])
y = df['income']
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(
  X,
  y,
  test_size = 0.20,
  random_state = 42,
  stratify=y         #we want the same ratio in both train and test...76% <=50K and 24% >50K
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler , OneHotEncoder
```

```python
numeric_cols = [
  "Age",
  "Education-num",
  "capital-gain",
  "capital-loss",
  "hours-per-week"
]

categorical_cols = [
  "Profession Class",
  "marital-status",
  "occupation",
  "relationship",
  #"race",
  "Gender",
  "country"
]
```

```python
#for numerical column
numeric_pipeline = Pipeline([
  ("imputer", SimpleImputer(strategy="median")),
  ("scaler", StandardScaler())
])
```

```python
#numeric_pipeline
```

```python
categorical_pipeline = Pipeline([
  ("imputer", SimpleImputer(strategy="most_frequent")),
  ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
categorical_pipeline
```

```python
preprocessor = ColumnTransformer([
  ("num", numeric_pipeline, numeric_cols),
  ("cat", categorical_pipeline, categorical_cols)
])
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = Pipeline([
  ("preprocessor" , preprocessor),
  ("model" , LogisticRegression(
    max_iter = 2000
  ))
])
logistic_model.fit(X_train, y_train)
```

```python
#now , for the prediction:
y_pred = logistic_model.predict(X_test)
```

### Evaluate it

```python
from sklearn.metrics import(
  accuracy_score,
  precision_score,
  recall_score,
  f1_score,  #for the minority class , thus making precision and recall balanced
  roc_auc_score,
  confusion_matrix,
  classification_report
)

print("accuracy : " , accuracy_score(y_test , y_pred))
#we have to pass the positive class label as precision and recall , f1 score need the positive label
#so u have to pass the positive label as in our case it's >50k....pos_label = 1
print("precision_score : " , precision_score(y_test , y_pred, pos_label='>50K'))
print("recall_score : " , recall_score(y_test , y_pred, pos_label='>50K'))
print("f1_score :", f1_score(y_test, y_pred, pos_label='>50K'))
print(logistic_model.classes_)
#predict the probability of the second class....so it will give the probability for both <=50k,>50k
y_prob = logistic_model.predict_proba(X_test)[:,1]
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
```

```python
#confusion_matrix
cm = confusion_matrix(y_test , y_pred)

sns.heatmap(
  cm,
  annot = True,
  cmap = 'viridis',
  fmt = '.2f'
)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = Pipeline([
  ("preprocessor" , preprocessor),
  ("model" , DecisionTreeClassifier(
    max_depth=10,
    min_samples_leaf=4,
    random_state = 42
  ))
])

dt_model.fit(X_train , y_train)
```

```python
y_pred_dt = dt_model.predict(X_test)

y_prob_dt = dt_model.predict_proba(X_test)[:,1]

print("Decision Tree Metrics")
print("accuracy: " , accuracy_score(y_test , y_pred_dt))
print("precision: " , precision_score(y_test , y_pred_dt , pos_label = '>50K'))
print("recall: " , recall_score(y_test , y_pred_dt , pos_label = '>50K'))
print("f1_score: " , f1_score(y_test , y_pred_dt , pos_label = '>50K'))
print("roc-auc: " , roc_auc_score(y_test , y_prob_dt))
```

```python
cm = confusion_matrix(y_test , y_pred_dt)

sns.heatmap(
  cm,
  annot = True,
  cmap = 'magma',
  fmt = '.2f'
)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title("Decision Tree Confusion Matrix")
plt.show()
```

**Random Forest**

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = Pipeline([
  ("preprocessor" , preprocessor),
  ("model" , RandomForestClassifier(
    n_estimators = 500,
    min_samples_leaf = 4,
    max_features = 'sqrt',
    #class_weight = "balanced", ->lowering the accuracy , precision , recall , roc-auc
    random_state = 42,
    n_jobs=-1
  )),
])
rf_model.fit(X_train , y_train)
```

Random Forest Beat Logistic Regression due to the ability of capturing non linear relationships....not like logReg..makes simple decision boundary..

```python
from sklearn.metrics import(
  accuracy_score,
  precision_score,
  recall_score,
  f1_score,
  roc_auc_score,
  confusion_matrix,
  classification_report
)

#make predictions using the Random Forest model
y_pred_rf = rf_model.predict(X_test)
# Predict probabilities using the Random Forest model
#y_prob_rf = rf_model.predict_proba(X_test)[:,1]

y_prob_rf = rf_model.predict_proba(X_test)[:,1]

print("Random Forest Model Metrics:")
print("accuracy : " , accuracy_score(y_test , y_pred_rf))
print("precision_score : " , precision_score(y_test , y_pred_rf, pos_label='>50K'))
print("recall_score : " , recall_score(y_test , y_pred_rf, pos_label='>50K'))
print("f1_score :", f1_score(y_test, y_pred_rf, pos_label='>50K'))
# print(rf_model.classes_) # The classes should be the same as logistic_model.classes_
print("ROC-AUC:", roc_auc_score(y_test, y_prob_rf))
```

```python
cm = confusion_matrix(y_test , y_pred_rf)

sns.heatmap(
  cm,
  annot = True,
  cmap = 'viridis',
  fmt = '.2f'
)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

### HyperParameter Tuning for Random Forst

```python
param_grid = {
  "model__n_estimators":[200,300,500],
  "model__max_depth":[None , 10 , 20],
  "model__min_samples_leaf":[1,2,4],
  #"model_max_features":["sqrt", "log2"]
}
```

```python
from sklearn.model_selection import RandomizedSearchCV

search = RandomizedSearchCV(
  rf_model,
  param_distributions = param_grid,
  n_iter = 10,
  scoring = "f1",
  #instead of one_train/validation set will make multiple folds
  cv=3,
  random_state = 42,
  n_jobs = -1
)

search.fit(X_train , y_train)
```

```python
print("Best Parameters:")
print(search.best_params_)

print("\nBest CV F1 Score:")
print(search.best_score_)
```

```python
#best tuned model nikalo
best_rf = search.best_estimator_
```

```python
y_pred_rf_tuned = best_rf.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

y_prob_rf_tuned = best_rf.predict_proba(X_test)[:,1]

print("Accuracy:", accuracy_score(y_test, y_pred_rf_tuned))
print("Precision:", precision_score(y_test, y_pred_rf_tuned , pos_label = '>50K'))
print("Recall:", recall_score(y_test, y_pred_rf_tuned , pos_label = '>50K'))
print("F1:", f1_score(y_test, y_pred_rf_tuned , pos_label = '>50K'))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_rf_tuned))
```
